```python
from langchain_huggingface import HuggingFacePipeline,ChatHuggingFace
```

```python
from transformers import AutoTokenizer
```

```python
import transformers
```

```python
import torch
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
model = "meta-llama/Llama-2-7b-chat-hf"
```

```python
tokenizer=AutoTokenizer.from_pretrained(model)
```

```python
pipe=transformers.pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    do_sample=True,
    temperature=0.5,
    top_k=10,
    max_new_tokens=500,
    return_full_text=False,
)
```

```python
llm=HuggingFacePipeline(pipeline=pipe)
chat_model = ChatHuggingFace(llm=llm)
```

```python
prompt ="What would be a good name for a company that makes colorful socks?"
```

```python
print(pipe.model.device)
```

```python
print(chat_model.invoke("Say hello in one sentence.").content)
```

```python
print(chat_model.invoke("What is a large language model?").content)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt_template1=PromptTemplate(input_variables=["cuisine"],template="I want to open a restaurant for {cuisine} food. Suggest a fance name for this")
```

```python
input_prompt=prompt_template1.format(cuisine="indian")
```

```python
print(input_prompt)
```

```python
prompt_template2=PromptTemplate(input_variables=["book_name"],template="Provide me a concise summary of the book {book_name}")
```

```python
input_prompt=prompt_template2.format(book_name="Alchemist")
```

```python
print(input_prompt)
```

```python
chain = prompt_template2 | chat_model
chain2 = prompt_template1 | chat_model
```

```python
response = chain2.invoke("indian")
print(response)
```
